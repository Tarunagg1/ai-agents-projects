```python
import os
from dotenv import load_dotenv

load_dotenv()


os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model= init_chat_model(model="gpt-4.1")

model
```

```python
response = model.invoke("Hello, how are you?")
response
```

```python
from langchain.chat_models import init_chat_model

modelgoogle = init_chat_model(model="google-genai:gemini-2.5-flash-lite")

# response = modelgoogle.invoke("Hello, how are you?")

genfrator = modelgoogle.stream("Write me a 200 word essay on the topic of 'The Importance of Education'")


for chunk in genfrator:
    print(chunk.text,end="|", flush=True)

```

```python
responses = modelgoogle.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
], config={
    "temperature": 0.7,
    "max_tokens": 100,
    "max_concurrency": 10
})
for response in responses:
    print(response)
```
