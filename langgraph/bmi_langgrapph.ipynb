```python
from langgraph.graph import StateGraph
from typing import TypedDict
```

```python
class BMMIState(TypedDict):
    weight: float
    height: float
    bmi: float
    category: str
```

```python
def calculate_bmi(state:BMMIState) -> BMMIState:
    state["bmi"] = state["weight"] / (state["height"] ** 2)
    return state
```

```python
def label_bmi(state: BMMIState) -> BMMIState:
    if state["bmi"] < 18.5:
        state["category"] = "Underweight"
    elif state["bmi"] < 25:
        state["category"] = "Normal"
    elif state["bmi"] < 30:
        state["category"] = "Overweight"
    else:
        state["category"] = "Obesity"
    return state
```

```python
graph = StateGraph(BMMIState)


# add nodes on graph

graph.add_node("calculate_bmi", calculate_bmi)
graph.add_node('label_bmi', label_bmi)



graph.add_edge(StateGraph.START, "calculate_bmi")
graph.add_edge("calculate_bmi", "label_bmi")
graph.add_edge("label_bmi", StateGraph.END)


workflow = graph.compile()
```
